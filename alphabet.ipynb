```python

# Old Turkic Unicode block info:
# https://unicode.org/charts/nameslist/n_10C00.html
# https://en.wikipedia.org/wiki/Old_Turkic_(Unicode_block)

# ... code chart:
# https://unicode.org/Public/18.0.0/charts/PDF/U10C00.pdf

# Unicode block interval: 
0x10c00, 0x10c48

CHARS = ''.join(chr(c) for c in range(0x10c00, 0x10c48 + 1))

unicode_info = defaultdict(dict)

for char in CHARS:
    name = unicodedata.name(char).split()
    variant, sound_value = name[-2:]
    assert variant in ["ORKHON", "YENISEI"]
    unicode_info[sound_value][variant] = char

#unicode_info
```

```python


with open("alphabet_table_wiki.txt") as alphadata:
    lines = list()
    for line in alphadata.readlines():
        if line.startswith('#'):
            continue
        else:
            lines.append(line.strip())
        
indexes = list()
for index, line in enumerate(lines):
    if line.startswith(">>"):
        indexes.append(index)
    


tables = list()
for start, stop in pairwise(indexes):
    tables.append(lines[start:stop])
tables.append(lines[stop:])

del indexes
```

```python

alphadata = dict()
for table in tables:
    name = ' '.join(table[0].split()[1:])
    letters = list()
    chunk = dict()
    count = 0
    for line in table[2:]:
        if not line:
            pass
        else:
            count += 1
            match count:
                case 1:
                    line = line.rstrip("\u200e")

                    # new
                    chars = dict()
                    memo = '' 
                    for char in line.split("\u200e"):
                        try:
                            name_data = unicodedata.name(char)
                        except TypeError: # happens if char is empty str 
                            continue
                        variant, char_name = unicodedata.name(char).split()[-2:]
                        if memo:
                            assert char_name == memo 
                        else:
                            memo = char_name 
                        chars[variant] = char
                    chunk['name'] = char_name 
                    chunk['chars'] = chars 
                    chunk['letters'] = line
                    # old
                    #chunk['letters'] = line.split("\u200e")
                    # 
                case 2:
                    chunk['IPA'] = re.findall(r"/(.*?)/", line)
                case 3:
                    line = [part.strip() for part in line.split(',')]
                    chunk['transliteration'] = line 
                    letters.append(chunk) 
                    chunk = dict()
                    count = 0
    alphadata[name] = letters 

#del tables

alphadata
```
