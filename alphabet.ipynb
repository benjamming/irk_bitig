```python

# Old Turkic Unicode block info:
# https://unicode.org/charts/nameslist/n_10C00.html
# https://en.wikipedia.org/wiki/Old_Turkic_(Unicode_block)

# ... code chart:
# https://unicode.org/Public/18.0.0/charts/PDF/U10C00.pdf

# Unicode block interval: 
0x10c00, 0x10c48

CHARS = ''.join(chr(c) for c in range(0x10c00, 0x10c48 + 1))

unicode_info = defaultdict(dict)

for char in CHARS:
    name = unicodedata.name(char).split()
    variant, sound_value = name[-2:]
    assert variant in ["ORKHON", "YENISEI"]
    unicode_info[sound_value][variant] = char

#unicode_info
```

```python
from itertools import pairwise

with open("alphabet_table_wiki") as alphadata:
    lines = list()
    for line in alphadata.readlines():
        if line.startswith('#'):
            continue
        else:
            lines.append(line.strip())
        
indexes = list()
for index, line in enumerate(lines):
    if line.startswith(">>"):
        indexes.append(index)
    

tables = list()
for start, stop in pairwise(indexes):
    tables.append(lines[start:stop])
tables.append(lines[stop:])

del indexes

tables
```

```python
alphadata = dict()
for table in tables:
    table_name = ' '.join(table[0].split()[1:])
    letters = list()
    alphadata[table_name] = letters 
    #record = dict()
    count = 0
    for line in table[1:]:
        if not line:
            continue 
        else:
            count += 1
            match count:
                case 1:
                    line = line.rstrip("\u200e")
                    chars = list()
                    for letter in line.split("\u200e"):
                        variant, char_name = unicodedata.name(letter).split()[-2:]
                        chars.append({'letter':letter, 'variant':variant, 'name':char_name})
                case 2:
                    for char in chars:
                        char['IPA'] = re.findall(r"/(.*?)/", line)
                case 3:
                    transliterations = [part.strip() for part in line.split(',')]
                    for char in chars:
                        char['tr1'] = transliterations
                    letters.extend(chars)
                    count = 0

alphadata['Vowels']
```
